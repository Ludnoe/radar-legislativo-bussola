```python
import requests
import pandas as pd
from datetime import date, timedelta

BASE_URL = "https://dadosabertos.camara.leg.br/api/v2"
HEADERS = {"Accept": "application/json"}

resp = requests.get(f"{BASE_URL}/partidos", headers=HEADERS, params={"itens": 10}, timeout=15)
print("Status HTTP:", resp.status_code)

json_bruto = resp.json()
print("Chaves principais da resposta:", list(json_bruto.keys()))
print("Links de paginação:", json_bruto["links"])

df_partidos = pd.DataFrame(json_bruto["dados"])
print("\nColunas disponíveis:", df_partidos.columns.tolist())
df_partidos.head()
```

```python
resp_dep = requests.get(f"{BASE_URL}/deputados", headers=HEADERS, params={"itens": 10}, timeout=15)
print("Status HTTP:", resp_dep.status_code)

df_deputados = pd.DataFrame(resp_dep.json()["dados"])
print("Colunas de deputados:", df_deputados.columns.tolist())
df_deputados.head()
```

```python
hoje = date.today()
sete_dias_atras = hoje - timedelta(days=7)

params_prop = {
    "dataApresentacaoInicio": sete_dias_atras.isoformat(),
    "dataApresentacaoFim": hoje.isoformat(),
    "itens": 10,
    "ordem": "DESC",
    "ordenarPor": "id"
}

resp_prop = requests.get(f"{BASE_URL}/proposicoes", headers=HEADERS, params=params_prop, timeout=15)
print("Status HTTP:", resp_prop.status_code)
print("Links:", resp_prop.json()["links"])

df_prop = pd.DataFrame(resp_prop.json()["dados"])
print("Qtd retornada nesta página:", len(df_prop))
print("Colunas de proposições:", df_prop.columns.tolist())
df_prop.head()
```

```python
# 1. Votações dos últimos 7 dias
params_vot = {
    "dataInicio": sete_dias_atras.isoformat(),
    "dataFim": hoje.isoformat(),
    "itens": 10
}
resp_vot = requests.get(f"{BASE_URL}/votacoes", headers=HEADERS, params=params_vot, timeout=15)
df_vot = pd.DataFrame(resp_vot.json()["dados"])
print("Status votações:", resp_vot.status_code)
print("Colunas de votações:", df_vot.columns.tolist())
display(df_vot.head(3))

# 2. Despesas de APENAS 1 deputado da lista que pegamos na Célula 2
primeiro_deputado_id = int(df_deputados.iloc[0]["id"])
resp_desp = requests.get(
    f"{BASE_URL}/deputados/{primeiro_deputado_id}/despesas",
    headers=HEADERS,
    params={"itens": 10},
    timeout=15
)
df_desp = pd.DataFrame(resp_desp.json()["dados"])
print("\nStatus despesas:", resp_desp.status_code, "| Deputado ID:", primeiro_deputado_id)
print("Colunas de despesas:", df_desp.columns.tolist())
display(df_desp.head(3))
```
